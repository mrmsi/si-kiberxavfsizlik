*"Sun'iy intellektni kiberxavfsizlikda qo'llash" darsligi, 7-bob.*

Quyidagi dasturda uch xil ma'lumotning entropiyasi taqqoslanadi: oddiy matn, siqilgan ma'lumot va tasodifiy baytlar (shifrlangan ma'lumotga o'xshash). So'ngra o'rtasiga "yashirin" shifrlangan qism qo'shilgan fayl bo'laklar bo'yicha tahlil qilinadi. Oxirida matndagi eng ko'p uchraydigan bayt juftliklari, ya'ni bayt 2-gramlari hisoblanadi. Dastur amaliy mashg'ulotning A7 mavzusiga tayyorgarlik bo'lib xizmat qiladi.

**Ishga tushirish:** yuqoridagi menyudan *Runtime → Run all* (*Ishga tushirish → Hammasini bajarish*) ni tanlang yoki kataklarni birma-bir Shift+Enter bilan bajaring. Barcha ma'lumotlar dasturning o'zida yaratiladi, qo'shimcha fayl yuklash shart emas.

````python
import math
import random
import zlib
from collections import Counter

tasodif = random.Random(0)   # natija har safar bir xil chiqishi uchun

def entropiya(baytlar):
    """Baytlar ketma-ketligining Shennon entropiyasi (0 dan 8 gacha)."""
    if not baytlar:
        return 0.0
    n = len(baytlar)
    return -sum(c / n * math.log2(c / n) for c in Counter(baytlar).values())
````

````python
# 1. Uch xil "fayl" yaratamiz
matn = ("Sun'iy intellektni kiberxavfsizlikda qo'llash fani. " * 400).encode()
siqilgan = zlib.compress(matn + tasodif.randbytes(2000))   # siqilgan (paketlangan) ma'lumot
shifrlangan = tasodif.randbytes(len(matn))                  # shifrlanganga o'xshash tasodifiy baytlar

for nom, b in [("Oddiy matn", matn), ("Siqilgan", siqilgan), ("Shifrlangan", shifrlangan)]:
    print(f"{nom:<12} hajmi {len(b):>6} bayt, entropiya {entropiya(b):.2f}")
````

````python
# 2. Fayl bo'laklari bo'yicha entropiya (odatiy qism + yashirilgan qism)
fayl = matn[:4096] + tasodif.randbytes(4096) + matn[:4096]
print("\nBo'laklar bo'yicha entropiya (har bir bo'lak 2048 bayt):")
for i in range(0, len(fayl), 2048):
    e = entropiya(fayl[i:i + 2048])
    print(f"  {i:>6}-bayt: {e:.2f} {'#' * int(e * 4)}")
````

````python
# 3. Bayt 2-gramlari: eng ko'p uchraydigan juftliklar
juftlar = Counter(matn[i:i + 2] for i in range(len(matn) - 1))
print("\nMatndagi eng ko'p uchraydigan bayt 2-gramlari:", [j.hex() for j, _ in juftlar.most_common(5)])
````
