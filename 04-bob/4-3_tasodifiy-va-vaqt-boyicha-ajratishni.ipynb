*"Sun'iy intellektni kiberxavfsizlikda qo'llash" darsligi, 4-bob.*

Quyidagi dasturda 12 oylik sun'iy ma'lumotlar yaratiladi. Har oy 500 ta fayl kuzatiladi, ulardan 10 foizi zararli. Zararli dasturlar har oy biroz o'zgaradi, ya'ni ularning xususiyatlari asta-sekin oddiy fayllarnikiga yaqinlashadi. Bu buzg'unchilarning antiviruslardan yashirinish uchun o'z dasturlarini doimiy takomillashtirib borishini modellashtiradi. Model avval tasodifiy ajratilgan ma'lumotlarda, so'ngra vaqt bo'yicha ajratilgan ma'lumotlarda baholanadi. Dastur amaliy mashg'ulotning A5 mavzusiga mos keladi.

**Ishga tushirish:** yuqoridagi menyudan *Runtime → Run all* (*Ishga tushirish → Hammasini bajarish*) ni tanlang yoki kataklarni birma-bir Shift+Enter bilan bajaring. Barcha ma'lumotlar dasturning o'zida yaratiladi, qo'shimcha fayl yuklash shart emas.

````python
import numpy as np
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score

rng = np.random.default_rng(1)

# 12 oylik sun'iy ma'lumot: har oyda 500 ta fayl, ulardan 10 foizi zararli.
# Zararli dasturlar har oy biroz "o'zgaradi" (xususiyatlari siljiydi).
X, y, oy = [], [], []
for m in range(12):
    oddiy = rng.normal(0, 1, (450, 5))
    zararli = rng.normal(3.0 - 0.2 * m, 1, (50, 5))   # vaqt o'tishi bilan oddiy fayllarga yaqinlashadi
    X += [oddiy, zararli]
    y += [0] * 450 + [1] * 50
    oy += [m] * 500
X, y, oy = np.vstack(X), np.array(y), np.array(oy)

model = RandomForestClassifier(n_estimators=100, random_state=0)

# 1-usul: tasodifiy ajratish (o'tmish va kelajak aralashib ketadi)
X_o, X_s, y_o, y_s = train_test_split(X, y, test_size=0.3, stratify=y, random_state=0)
model.fit(X_o, y_o)
print("Tasodifiy ajratish:   F1 =", round(f1_score(y_s, model.predict(X_s)), 2))

# 2-usul: vaqt bo'yicha ajratish (dastlabki 6 oyda o'qitib, keyingi oylarda sinaymiz)
model.fit(X[oy < 6], y[oy < 6])
for m in range(6, 12):
    t = oy == m
    print(f"Vaqt bo'yicha, {m+1:>2}-oy: F1 =", round(f1_score(y[t], model.predict(X[t])), 2))
````
