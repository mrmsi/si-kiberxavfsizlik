Quyidagi dastur har bir URL manzil uchun sakkizta xususiyatni hisoblaydi. Dastur amaliy mashg'ulotning A4 mavzusiga mos keladi.

**Ishga tushirish:** yuqoridagi menyudan *Runtime → Run all* (*Ishga tushirish → Hammasini bajarish*) ni tanlang yoki kataklarni birma-bir Shift+Enter bilan bajaring. Barcha ma'lumotlar dasturning o'zida yaratiladi, qo'shimcha fayl yuklash shart emas.

````python
import math
import re
from collections import Counter
from urllib.parse import urlparse

def entropiya(matn):
    """Shennon entropiyasi: belgilar qanchalik tartibsiz taqsimlanganini o'lchaydi."""
    soni = Counter(matn)
    n = len(matn)
    return -sum(c / n * math.log2(c / n) for c in soni.values())

shubhali_sozlar = ["login", "verify", "secure", "account", "bank", "update"]

def url_xususiyatlari(url):
    q = urlparse(url)
    xost = q.netloc
    return {
        "uzunlik": len(url),
        "raqamlar": sum(ch.isdigit() for ch in url),
        "nuqtalar": xost.count("."),
        "defislar": xost.count("-"),
        "ip_manzil": int(bool(re.fullmatch(r"\d+\.\d+\.\d+\.\d+", xost))),
        "https": int(q.scheme == "https"),
        "shubhali_soz": sum(s in url.lower() for s in shubhali_sozlar),
        "entropiya": round(entropiya(xost), 2),
    }

manzillar = [
    "https://my.gov.uz/oz",
    "https://tsue.uz/talabalar",
    "http://secure-login-bank.uz-verify.top/account/update",
    "http://185.22.31.4/namunabank/login.php",
    "https://xk3j9q2z7vb.ru/a.php?id=88412",
]
for url in manzillar:
    print(url)
    print("   ", url_xususiyatlari(url))
````

## Natijalar tahlili
 Dastlabki ikki manzil haqiqiy davlat va universitet saytlariga tegishli: ular qisqa, raqamsiz, xavfsiz HTTPS protokolidan foydalanadi va shubhali so'zlarni o'z ichiga olmaydi. Uchinchi manzil bank saytiga o'xshatishga urinilgan: unda oltita shubhali so'z, uchta defis bor, domen nomida ".uz" qo'shimchasi boshqa domen ichiga yashirilgan, xost nomining entropiyasi esa yuqori. To'rtinchi manzilda domen nomi o'rniga to'g'ridan-to'g'ri IP manzil ishlatilgan, bu fishing saytlarga xos belgi hisoblanadi. Beshinchi manzil zararli dastur tomonidan avtomatik yaratilgan domenga o'xshaydi: tasodifiy belgilar ketma-ketligi va yuqori entropiya bunga ishora qiladi. E'tibor bering, beshinchi manzil HTTPS protokolidan foydalanadi. Demak, "HTTPS bo'lsa, sayt xavfsiz" degan keng tarqalgan fikr noto'g'ri: bugungi kunda ko'plab fishing saytlar ham HTTPS sertifikatiga ega.
