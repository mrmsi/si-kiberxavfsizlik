## Amaliy misol: "99 foiz aniq" tizim

[![Colab'da ochish](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/mrmsi/si-kiberxavfsizlik/blob/main/05-bob/5-2_99-foiz-aniq-tizim.ipynb)

*"Sun'iy intellektni kiberxavfsizlikda qo'llash" darsligi, 5-bob.*

Faraz qilaylik, bostirib kirishni aniqlash tizimi hujumlarning 99 foizini aniqlaydi, oddiy ulanishlarning esa atigi 1 foizini xato ravishda hujum deb belgilaydi. Tarmoqda kuniga bir million ulanish bo'ladi, ulardan har o'n mingtasidan bittasi hujum. Quyidagi dastur bu tizim amalda qanday natija berishini hisoblaydi.

**Ishga tushirish:** yuqoridagi menyudan *Runtime → Run all* (*Ishga tushirish → Hammasini bajarish*) ni tanlang yoki kataklarni birma-bir Shift+Enter bilan bajaring. Barcha ma'lumotlar dasturning o'zida yaratiladi, qo'shimcha fayl yuklash shart emas.

````python
# Bazaviy chastota xatosi: "99 foiz aniq" IDS amalda qanchalik foydali?
ulanishlar = 1_000_000          # bir kunlik ulanishlar soni
hujum_ulushi = 0.0001           # har 10 000 ulanishdan bittasi hujum
tpr = 0.99                      # to'liqlik: hujumlarning 99 foizini topadi
fpr = 0.01                      # oddiy ulanishlarning 1 foizini xato belgilaydi

hujumlar = ulanishlar * hujum_ulushi
oddiylar = ulanishlar - hujumlar
tp = hujumlar * tpr             # to'g'ri aniqlangan hujumlar
fp = oddiylar * fpr             # noto'g'ri signallar

print(f"Haqiqiy hujumlar:             {hujumlar:>8,.0f}")
print(f"To'g'ri aniqlangan hujumlar:  {tp:>8,.0f}")
print(f"Noto'g'ri signallar:          {fp:>8,.0f}")
print(f"Ogohlantirishlar jami:        {tp + fp:>8,.0f}")
print(f"Aniqlik: {tp / (tp + fp):.1%}")
````
