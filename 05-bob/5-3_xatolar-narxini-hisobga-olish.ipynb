*"Sun'iy intellektni kiberxavfsizlikda qo'llash" darsligi, 5-bob.*

Noto'g'ri signal va o'tkazib yuborilgan hujumning narxi turlicha. Masalan, bankda karta operatsiyasini xato ravishda bloklash mijozning noroziligiga va qo'ng'iroq markaziga qo'shimcha yuklamaga olib keladi. Firibgarlik operatsiyasini o'tkazib yuborish esa bevosita moliyaviy yo'qotishni anglatadi. Xavfsizlik bo'limi ushbu narxlarni taxminan baholab, chegarani umumiy narxni eng kam qiladigan qilib tanlashi mumkin.

**Ishga tushirish:** yuqoridagi menyudan *Runtime → Run all* (*Ishga tushirish → Hammasini bajarish*) ni tanlang yoki kataklarni birma-bir Shift+Enter bilan bajaring. Barcha ma'lumotlar dasturning o'zida yaratiladi, qo'shimcha fayl yuklash shart emas.

````python
import numpy as np
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix

rng = np.random.default_rng(3)
X = np.vstack([rng.normal(0, 1, (9800, 4)), rng.normal(1.5, 1, (200, 4))])
y = np.array([0] * 9800 + [1] * 200)          # 2 foiz hujum
X_o, X_s, y_o, y_s = train_test_split(X, y, test_size=0.5, stratify=y, random_state=0)

model = RandomForestClassifier(n_estimators=200, random_state=0).fit(X_o, y_o)
ehtimol = model.predict_proba(X_s)[:, 1]       # har bir ulanishning hujum ehtimoli

# Xato narxlari: o'tkazib yuborilgan hujum 50 birlik, noto'g'ri signal 1 birlik
narx_fn, narx_fp = 50, 1

print(f"{'Chegara':>8}{'TP':>6}{'FN':>6}{'FP':>6}{'Aniqlik':>9}{'To‘liqlik':>10}{'Narx':>7}")
for chegara in [0.1, 0.2, 0.3, 0.5, 0.7]:
    p = (ehtimol >= chegara).astype(int)
    tn, fp, fn, tp = confusion_matrix(y_s, p).ravel()
    aniqlik = tp / (tp + fp) if tp + fp else 0
    toliqlik = tp / (tp + fn)
    narx = fn * narx_fn + fp * narx_fp
    print(f"{chegara:>8.1f}{tp:>6}{fn:>6}{fp:>6}{aniqlik:>9.2f}{toliqlik:>10.2f}{narx:>7}")
````
