## Amaliy misol: xodim faoliyatidagi anomaliyalarni aniqlash

[![Colab'da ochish](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/mrmsi/si-kiberxavfsizlik/blob/main/03-bob/3-2_xodim-faoliyatidagi-anomaliyalarni-aniqlash.ipynb)

*"Sun'iy intellektni kiberxavfsizlikda qo'llash" darsligi, 3-bob.*

Quyidagi dasturda xodimning 60 kunlik faoliyati ikki xususiyat bilan tavsiflanadi: tizimga kirish soati va kun davomida yuklab olingan fayllar soni. Ma'lumotlarga ikkita shubhali kun qo'shilgan: birida xodim tunda tizimga kirgan, ikkinchisida esa odatdagidan 25 barobar ko'p fayl yuklab olgan. Izolyatsiyalovchi o'rmon hech qanday belgisiz shu kunlarni aniqlashi kerak. Dastur amaliy mashg'ulotning A10 va A11 mavzulariga mos keladi.

**Ishga tushirish:** yuqoridagi menyudan *Runtime → Run all* (*Ishga tushirish → Hammasini bajarish*) ni tanlang yoki kataklarni birma-bir Shift+Enter bilan bajaring. Barcha ma'lumotlar dasturning o'zida yaratiladi, qo'shimcha fayl yuklash shart emas.

````python
import numpy as np
from sklearn.ensemble import IsolationForest

rng = np.random.default_rng(7)

# Xodimning 60 kunlik faoliyati: tizimga kirish soati va yuklab olingan fayllar soni
kirish_soati = rng.normal(9, 0.7, 60)          # odatda soat 9 atrofida
fayllar      = rng.poisson(12, 60)              # kuniga 10-15 ta fayl
kunlar = np.column_stack([kirish_soati, fayllar])

# Ikki shubhali kun: tungi kirish va ko'p fayl yuklab olish
shubhali = np.array([[2.5, 14], [9.2, 340]])
barcha = np.vstack([kunlar, shubhali])

# Belgilar talab qilinmaydi: model "odatiy" holatni o'zi o'rganadi
model = IsolationForest(contamination=0.03, random_state=0).fit(barcha)
natija = model.predict(barcha)          # 1 - odatiy, -1 - anomaliya

for i in np.where(natija == -1)[0]:
    print(f"{i+1}-kun: kirish soati {barcha[i,0]:.1f}, fayllar {barcha[i,1]:.0f} -> anomaliya")
````
