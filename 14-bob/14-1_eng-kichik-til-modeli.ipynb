Quyidagi dasturda katta til modelining asosiy g'oyasi juda soddalashtirilgan ko'rinishda namoyish etiladi. Kiberxavfsizlik haqidagi 11 ta jumladan iborat kichik matn asosida **bigram modeli** quriladi: har bir so'zdan keyin qaysi so'z necha marta kelgani sanaladi. So'ngra model har qadamda keyingi so'zni shu ehtimollarga ko'ra tanlab, yangi jumlalar yaratadi. Haqiqiy katta til modellari ancha murakkab: ular oldingi bitta so'zga emas, minglab tokenlarga qaraydi va milliardlab parametrga ega. Ammo asosiy tamoyil, ya'ni keyingi elementni ehtimollar asosida tanlash, bir xil. Dastur amaliy mashg'ulotning A14 mavzusiga mos keladi.

**Ishga tushirish:** yuqoridagi menyudan *Runtime → Run all* (*Ishga tushirish → Hammasini bajarish*) ni tanlang yoki kataklarni birma-bir Shift+Enter bilan bajaring. Barcha ma'lumotlar dasturning o'zida yaratiladi, qo'shimcha fayl yuklash shart emas.

````python
import random
from collections import Counter, defaultdict

# Kichik o'quv matni (haqiqiy til modellari trillionlab so'zda o'qitiladi)
matn = """
fishing xati foydalanuvchini soxta saytga yo'naltiradi .
fishing xati parolni o'g'irlash uchun yuboriladi .
soxta sayt bank saytiga o'xshaydi .
zararli dastur parolni o'g'irlaydi .
zararli dastur fayllarni shifrlaydi .
to'lov talab qiluvchi dastur fayllarni shifrlaydi va pul talab qiladi .
antivirus zararli dasturni aniqlaydi .
model fishing xatini aniqlaydi .
model zararli dasturni aniqlaydi .
tahlilchi signalni tekshiradi .
tahlilchi fishing xatini tekshiradi .
"""
sozlar = matn.split()

# Bigram modeli: har bir so'zdan keyin qaysi so'z necha marta kelgan
keyingi = defaultdict(Counter)
for a, b in zip(sozlar, sozlar[1:]):
    keyingi[a][b] += 1

def ehtimollar(soz):
    jami = sum(keyingi[soz].values())
    return {k: round(v / jami, 2) for k, v in keyingi[soz].most_common()}

for s in ["zararli", "dastur", "xati"]:
    print(f"'{s}' so'zidan keyin: {ehtimollar(s)}")

# Matn yaratish: har qadamda keyingi so'zni ehtimollarga ko'ra tanlash
tasodif = random.Random(7)
print("\nModel yaratgan jumlalar:")
for _ in range(6):
    soz, jumla = tasodif.choice(["fishing", "zararli", "model", "tahlilchi", "antivirus"]), []
    while soz != "." and len(jumla) < 12:
        jumla.append(soz)
        variantlar = keyingi[soz]
        soz = tasodif.choices(list(variantlar), weights=list(variantlar.values()))[0]
    print("  ", " ".join(jumla))
````
